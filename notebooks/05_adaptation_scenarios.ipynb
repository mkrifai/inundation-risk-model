# Tutorial 5: Adaptation Scenarios & Spatial Planning Conflicts

Simulating climate change intensification (+15% rain), Nature-based Solutions (NbS), and spatial planning conflicts using inundation_risk.scenarios.

## 1. Imports

```python
import numpy as np
import matplotlib.pyplot as plt
from inundation_risk.scenarios import simulate_adaptation_scenario, evaluate_spatial_plan_conflicts
```

## 2. Simulate Adaptation Scenario

```python
rng = np.random.default_rng(42)
baseline_h = rng.uniform(0.1, 0.8, size=(100, 100))
baseline_r = baseline_h * 0.5

scen_h, scen_r, metrics = simulate_adaptation_scenario(
    baseline_hazard=baseline_h,
    baseline_risk=baseline_r,
    rainfall_multiplier=1.15,
    mitigation_efficiency=0.25
)
print('Scenario Results:', metrics)
```

## 3. Evaluate Spatial Planning Conflicts

```python
zoning = rng.choice([1, 2, 3, 4], size=(100, 100))
pop = rng.uniform(10, 500, size=(100, 100))

conflicts = evaluate_spatial_plan_conflicts(
    hazard_raster=scen_h,
    zoning_raster=zoning,
    planned_urban_codes=[1],
    population_density=pop,
    hazard_threshold=0.60
)
print('Spatial Planning Conflict Metrics:', conflicts)
```
